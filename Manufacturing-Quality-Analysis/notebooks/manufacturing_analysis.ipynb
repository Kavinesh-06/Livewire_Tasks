## **key quality KPIs**
### *KPI Definition*

1. **Defect rate** : Events with a recorded defect $\div$ total events $\times$ 100
2. **First-pass yield (FPY)** : Events passing the first inspection $\div$ total events $\times$ 100
3. **Scrap rate** : Scrapped events $\div$ total events $\times$ 100
4. **Rework rate** : Events requiring rework $\div$ total events $\times$ 100
5. **Average cycle time** : Mean total_cycle_time_min
6. **Total production cost** : Sum of cost_usd
7. **Warranty claim rate** : Events with a warranty claim $\div$ total events $\times$ 100

```python
total_events = len(df)

defect_events = df["defect_type"].ne("none").sum()
scrap_events = df["scrap"].eq(1).sum()
rework_events = df["decision_rework"].eq(1).sum()
final_pass_events = df["final_pass"].eq(1).sum()
warranty_events = df["warranty_claim_90d"].eq(1).sum()

kpis = pd.Series({
    "Total events": total_events,
    "Defect rate (%)": defect_events / total_events * 100,
    "Scrap rate (%)": scrap_events / total_events * 100,
    "Rework rate (%)": rework_events / total_events * 100,
    "Final pass rate (%)": final_pass_events / total_events * 100,
    "Average cycle time (min)": df["total_cycle_time_min"].mean(),
    "Total recorded cost (USD)": df["cost_usd"].sum(),
    "Warranty claim rate (%)": warranty_events / total_events * 100
})

display(kpis.round(2))
```

## 1. Explore quality patterns

```python
# Which defect types occur most often

defect_counts = (
    df.loc[df["defect_type"] != "none", "defect_type"]
    .value_counts()
)

display(defect_counts)

ax = defect_counts.plot(
    kind="bar",
    figsize=(9, 5),
    title="Manufacturing Defects by Type"
)
ax.bar_label(ax.containers[0], padding=3)
plt.xlabel("Defect Type")
plt.ylabel("Number of Events")
plt.xticks(rotation=45, ha="right")
plt.ylim(0,400)
plt.tight_layout()
plt.show()
```

```python
# Which production lines have the highest defect rate

line_quality = (
    df.assign(
        has_defect=df["defect_type"].ne("none")
    )
    .groupby("line")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum")
    )
)

line_quality["defect_rate_pct"] = (
    line_quality["defective_events"]
    / line_quality["total_events"] * 100
)

display(
    line_quality.sort_values(
        "defect_rate_pct", ascending=False
    ).round(2)
)
```

```python
# defect rate differ by shift

shift_quality = (
    df.assign(has_defect=df["defect_type"].ne("none"))
    .groupby("shift")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum")
    )
)

shift_quality["defect_rate_pct"] = (
    shift_quality["defective_events"]
    / shift_quality["total_events"] * 100
)

display(shift_quality.sort_values(
        "defect_rate_pct", ascending=False).round(2))
```

## 2. Machine performance :-

```python
# Create a consistent defect indicator
df["has_defect"] = (
    df["defect_type"].astype("string")
    .str.strip()
    .str.lower()
    .ne("none")
    .fillna(False)
)

# Convert the indicator to a regular Boolean column
df["has_defect"] = df["has_defect"].astype(bool)

print("Dataset shape:", df.shape)
print("Defective events:", df["has_defect"].sum())
print("Overall defect rate (%):",
      round(df["has_defect"].mean() * 100, 2))
```

#### **Compare machines**

```python
machine_quality = (
    df.groupby("machine_id")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum"),
        avg_machine_age=("machine_age_yrs", "mean"),
        avg_cycle_time=("total_cycle_time_min", "mean"),
        scrap_events=("scrap", "sum")
    )
)

machine_quality["defect_rate_pct"] = (
    machine_quality["defective_events"]
    / machine_quality["total_events"] * 100
)

machine_quality["scrap_rate_pct"] = (
    machine_quality["scrap_events"]
    / machine_quality["total_events"] * 100
)

# Focus on machines with at least 30 recorded events
machine_quality_filtered = machine_quality.query(
    "total_events >= 30"
)

display(
    machine_quality_filtered
    .sort_values("defect_rate_pct", ascending=False)
    .round(2)
    .head(10)
)
print("Total machines:", len(machine_quality))
```

#### **Investigate machine age**

```python

df["machine_age_group"] = pd.cut(
    df["machine_age_yrs"],
    bins=[-np.inf, 2, 5, 8, np.inf],
    labels=["0–2 years", "2–5 years",
            "5–8 years", "8+ years"],
    right=False
)

age_quality = (
    df.groupby("machine_age_group", observed=True)
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum"),
        avg_cycle_time=("total_cycle_time_min", "mean")
    )
)

age_quality["defect_rate_pct"] = (
    age_quality["defective_events"]
    / age_quality["total_events"] * 100
)
display(age_quality.round(2))

ax = age_quality["defect_rate_pct"].plot(
    kind="bar",
    figsize=(9, 6),
    title="Defect Rate by Machine Age"
)
ax.bar_label(ax.containers[0],fmt='%.2f',padding=5)

plt.xlabel("Machine Age")
plt.ylabel("Defect Rate (%)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 3. Material quality :-

#### **Calculate quality metrics by material and the most common defects within each grade.**

```python

material_quality = (
    df.groupby("material_grade")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum"),
        scrap_events=("scrap", "sum"),
        avg_cost_usd=("cost_usd", "mean")
    )
)

material_quality["defect_rate_pct"] = (
    material_quality["defective_events"]
    / material_quality["total_events"] * 100
)

material_quality["scrap_rate_pct"] = (
    material_quality["scrap_events"]
    / material_quality["total_events"] * 100
)

display(
    material_quality
    .sort_values("defect_rate_pct", ascending=False)
    .round(2)
)

material_defects = pd.crosstab(
    df["material_grade"],
    df.loc[df["has_defect"], "defect_type"]
)

display(material_defects)
```

#### **Compare material grades**

```python

plt.figure(figsize=(7, 6))

ax = material_quality["defect_rate_pct"].plot(
    kind="bar",
    figsize=(9, 6),
    title="Defect Rate by Machine grade"
)
ax.bar_label(ax.containers[0],fmt='%.1f',padding=3)

plt.title("Defect Rate by Material Grade")
plt.xlabel("Material Grade")
plt.ylabel("Defect Rate (%)")
plt.tight_layout()
plt.show()
```

## 4. Process conditions :-
*By temperature, humidity, and production speed relate to defect rates*

#### **Temperature analysis**

```python

df["temperature_band"] = pd.cut(
    df["temp_c"],
    bins=[-np.inf, 10, 20, 30, np.inf],
    labels=["Below 10°C", "10–20°C",
            "20–30°C", "Above 30°C"]
)

temperature_quality = (
    df.groupby("temperature_band", observed=True)
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum")
    )
)

temperature_quality["defect_rate_pct"] = (
    temperature_quality["defective_events"]
    / temperature_quality["total_events"] * 100
)

display(temperature_quality.round(2))
```

#### **Humidity analysis**

```python

df["humidity_band"] = pd.cut(
    df["humidity_pct"],
    bins=[-np.inf, 40, 60, 80, np.inf],
    labels=["Below 40%", "40–60%",
            "60–80%", "Above 80%"]
)

humidity_quality = (
    df.groupby("humidity_band", observed=True)
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum")
    )
)

humidity_quality["defect_rate_pct"] = (
    humidity_quality["defective_events"]
    / humidity_quality["total_events"] * 100
)

display(humidity_quality.round(2))
```

#### **production-speed Analysis**

```python
df["production_speed_band"] = pd.qcut(
    df["process_speed_units_hr"],
    q=4,
    duplicates="drop"
)
production_speed_quality = (
    df.groupby(
        "production_speed_band",observed=True
    )
    .agg(
        total_events = ("event_id","count"),
        defect_events = ("has_defect","sum"),
        avg_cycle_time = ("total_cycle_time_min","mean")
    )
)

production_speed_quality["defect_rate_pct"] = (
    production_speed_quality["defect_events"]
    / production_speed_quality["total_events"] * 100
)
display(production_speed_quality.round(2))
```

#### **Plot defect rates across the three factors**
*Tempereature, Humidity, and Production speed.*

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 6))

temperature_quality["defect_rate_pct"].plot(
    kind="bar", ax=axes[0], title="Defect Rate by Temperature"
)
humidity_quality["defect_rate_pct"].plot(
    kind="bar", ax=axes[1], title="Defect Rate by Humidity"
)
production_speed_quality["defect_rate_pct"].plot(
    kind="bar", ax=axes[2], title="Defect Rate by Speed"
)

for ax in axes:
    ax.set_ylabel("Defect Rate (%)")
    ax.tick_params(axis="x", rotation=45)
    ax.bar_label(ax.containers[-1],fmt="%.2f",padding=3)

plt.tight_layout()
plt.show()
```

## 5.Cost analysis :-
*recorded costs differ between defective, reworked, and scrapped events*
#### **Compare cost by quality outcome**

```python

df["quality_outcome"] = np.select(
    [df["scrap"].eq(1), df["decision_rework"].eq(1), df["has_defect"]],
    ["Scrapped", "Reworked", "Defective - no rework/scrap flag"],
    default="No recorded defect"
)

cost_analysis = (
    df.groupby("quality_outcome")
    .agg(
        total_events=("event_id", "count"),
        total_recorded_cost=("cost_usd", "sum"),
        avg_recorded_cost=("cost_usd", "mean"),
        median_recorded_cost=("cost_usd", "median"),
        avg_cycle_time=("total_cycle_time_min", "mean"),
        avg_rework_time=("rework_time_min", "mean")
    )
)

display(cost_analysis.round(2))
```

#### **Estimate the cost associated with scrap and rework**

```python
scrap_cost_summary = df.loc[df["scrap"].eq(1)].agg(
    events=("event_id", "count"),
    total_recorded_cost=("cost_usd", "sum"),
    avg_recorded_cost=("cost_usd", "mean")
)

rework_cost_summary = df.loc[df["decision_rework"].eq(1)].agg(
    events=("event_id", "count"),
    total_recorded_cost=("cost_usd", "sum"),
    avg_recorded_cost=("cost_usd", "mean")
)

print("Scrap summary:")
display(scrap_cost_summary)

print("Rework summary:")
display(rework_cost_summary)
```

## 6. Time trends :-
*defects becoming more frequent over time*
#### **Create daily defect metrics**

```python

valid_dates = df.dropna(subset=["event_ts"]).copy()

daily_quality = (
    valid_dates
    .set_index("event_ts")
    .resample("D")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum")
    )
)

daily_quality["defect_rate_pct"] = (
    daily_quality["defective_events"]
    / daily_quality["total_events"].replace(0, np.nan) * 100
)

display(daily_quality.head(10))
```

#### **Analyze monthly trends**

```python

monthly_quality = (
    valid_dates
    .set_index("event_ts")
    .resample("MS")
    .agg(
        total_events=("event_id", "count"),
        defective_events=("has_defect", "sum"),
        scrap_events=("scrap", "sum")
    )
)

monthly_quality["defect_rate_pct"] = (
    monthly_quality["defective_events"]
    / monthly_quality["total_events"] * 100
)

monthly_quality["scrap_rate_pct"] = (
    monthly_quality["scrap_events"]
    / monthly_quality["total_events"] * 100
)

display(monthly_quality.round(2))

monthly_quality["defect_rate_pct"].plot(
    marker="o",
    figsize=(11, 5),
    title="Monthly Manufacturing Defect Rate"
)

plt.xlabel("Month")
plt.ylabel("Defect Rate (%)")
plt.tight_layout()
plt.show()
```

## 7. Defect severity
*defect types have the highest severity, and which require the most attention. where higher values indicate greater severity according to the field's, severity **2–3 as an initial high-severity** grouping for exploration*

#### **Identify high-severity defects**

```python
severity_counts = df["defect_severity_0to3"].value_counts(
    dropna=False
).sort_index()

display(severity_counts)

high_severity = df.loc[
    df["has_defect"]
    & df["defect_severity_0to3"].ge(2)
].copy()

high_severity_summary = (
    high_severity.groupby("defect_type")
    .agg(
        event_count=("event_id", "count"),
        avg_severity=("defect_severity_0to3", "mean"),
        scrap_events=("scrap", "sum"),
        rework_events=("decision_rework", "sum"),
        avg_recorded_cost=("cost_usd", "mean")
    )
    .sort_values(
        ["event_count", "avg_severity"],
        ascending=False
    )
)
display(high_severity_summary.round(2))
```

#### **the consistency of severity and defect labels**

```python
top_severe_defects = (
    high_severity["defect_type"]
    .value_counts()
    .sort_values()
)
display(top_severe_defects)
severity_check = pd.crosstab(
    df["defect_type"],
    df["defect_severity_0to3"],
    margins=True
)

display(severity_check)
```
