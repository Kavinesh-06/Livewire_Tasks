#### 2. Reverse a String

```python
text = input(("Enter text : "))
reverse = ''
for i in range(len(text)-1,-1,-1):
    reverse += text[i]
print(reverse)
```

#### 3. Palindrome String

```python
text = input("Enter text : ")
reverse = text[::-1]
if text == reverse:
    print(f"The given text {text} is Palindrome !")
else:
    print(f"The given text {text} is not an Palindrome !")
```

#### 4. Count Vowels

#### 5. Count Consonants

#### 12. Longest Word

```python
text = input("Enter the sentence : ")
result = text.split()
n = len(result)
large = result[0]
for i in range(1,n):
    if len(result[i]) >= len(large):
        large = result[i]
print(large)
```

#### 14. Character Frequency

```python
text = input("Enter the sentence : ")
result = ''
ans = {}
for i in text:
    ans[i] = 1
    if i in result:
       ans[i] += 1
    result += i
print(f"Character Frequency {ans}")
```

#### 17. Anagram Check

```python
text = input("Enter some texts : ")
text_1 = input("Enter some other texts : ")
check = ''
if len(text) == len(text_1):
    for i in text:
        if i in text_1:
            check += i
    if check == text:
        print("Anagram")
    else:
        print("Not an Anagram")
else:
    print("Not an Anagram")
```

#### 21. String Rotation

```python
text = input("Enter string_1 : ").upper()
text_1 = input("Enter string_2 : ").upper()
n = len(text)
a = False
for i in range (1,n+1):
    rotation = text[i:] + text[:i]
    if text_1 == rotation:
        print(f"{text_1} is present in the rotation of {text}")
        a = True
if not a:
    print(f"{text_1} is not present in the rotation of {text}")
```

#### 23. String Compression

```python
text = input("Enter string : ")
x =''
count = 1
for i in range(len(text)):
        if i < len(text)-1 and text[i] == text[i+1]:
            count += 1
        else:
            x += text[i] + str(count)
            count = 1
print(x)
```

#### 24. Password Validation

```python
print("""Minimum 8 characters in length
At least one uppercase letter
At least one lowercase letter
At least one digit
""")
text = input("Enter valid passord : ")
upper_case, lower_case,digit = False,False,False
for i in text:
    if len(text) >= 8:
        if i.isupper():
            upper_case = True
        elif i.islower():
            lower_case = True
        elif i.isdigit():
            digit = True
    else:
        print("Your passwords must be minimum 8 characters in length")
if upper_case and lower_case and digit:
    print("valid password")
else:
    print("Invalid password")
```

#### 25. Real-Time Email Validation

```python
text = input("Enter an email address : ")
username = text.split('@')[0]
if text.islower() and username.isalpha() and ('@' in text) and (text.index('@')) < (text.index('.')) and text.endswith("@gmail.com"):
    print(f"{text} → Valid Email")
else:
    print(f"{text} → Invalid Email")
```
