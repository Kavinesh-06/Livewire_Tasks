#### 10. Sum of Digits

```python
number = int(input("Enter numbers : "))
value = 0
while number>0:
    remain = number%10
    value += remain
    number //= 10
print(value)
```

#### 14. Prime Number Check

```python
n = int(input("Enter an value : "))
count = 0
for i in range(2,n//2 + 1):
    if n % i == 0:
        count = 1
if count == 0:
    print("yes!, it's a prime number")
else:
    print("Its' not a prime number")
```

#### 15. Prime Numbers up to N

```python
n = int(input("Enter an value : "))
for j in range(2,n+1):
    count = 0
    for i in range(2,j//2 + 1):
        if j % i == 0:
            count = 1
    if count == 0:
        print(j,end=' ')
```

#### 16. Greatest Common Factor (GCF/GCD)

```python
a = int(input("Enter a value : "))
b = int(input("Enter a value : "))
low = min(a,b)
gcd = 1
for i in range(1,low+1):
    if a%i == 0 and b%i == 0 :
        gcd = i
print(gcd)
```

#### 19. Factorial Calculation 

```python
num = int(input("Enter some value : "))
fact = 1
if num == 0:
    print(f" The factorial of a {num} is 0.")
else:
    for i in range(1,num+1):
        fact *= i
    print(f" The factorial of a {num} is {fact}.")
```

#### 20. Sum of Even and Odd Numbers 

```python
x = int(input("Enter an value : "))
odd = 0
even = 0
for i in range(1,x+1):
    if i%2==0:
        even += i
    else:
        odd += i
print(f"The sum of even numbers is {even}.")
print(f"The sum of odd number is {odd}.")
```

#### 21. Palindrome Number Check

```python
num = int(input("Enter a number : "))
num_1 = num
rev = 0
while num>0:
    rem = num % 10
    rev = rev*10 + rem
    num //= 10 
if rev == num_1 :
    print(f"input = {num_1} and Reverse = {rev} (palindrome)")
else:
    print(f"input = {num_1} and Reverse = {rev} (Not palindrome)")
```

#### 22. Armstrong Number Check

```python
num = int(input("Enter an Value : "))
num_1 = num
a = 0
while(num > 0):
    a = (num % 10 )**3 + a
    num //= 10
if a == num_1:
    num_1=str(num_1)
    print(f"{num_1} is an Armstrong number because {num_1[0]}^3 + {num_1[1]}^3 + {num_1[2]}^3 = {a}.")
else:
    num_1=str(num_1)
    print(f"{num_1} is not an Armstrong number because {num_1[0]}^3 + {num_1[1]}^3 + {num_1[2]}^3 = {a}.")
```

#### 23. Pattern Printing Practice 

**Pattern A (Right-Angled Triangle):**

```python
for i in range(5):
    for j in range(i+1):
        print('*',end=' ')
    print('\n')
```

**Pattern B (Number Pyramid):**

```python
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end=' ')
    print('\n')
```
