#### *Task 1: Line and Word Counter*

```python
file = open('document.txt','r')
load = file.readlines()
length = 0
for i in load:
    words = i.split()
    #print(words)
    length += len(words)

#print(load)
print(f"Total Lines: {len(load)} Total Words: {length}")
```

#### *Task 2: User Note Writer*
