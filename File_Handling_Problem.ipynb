#### *Task 1: Line and Word Counter*

```python
file = open('documents/document.txt','r')
load = file.readlines()
length = 0
for i in load:
    words = i.split()
    #print(words)
    length += len(words)

# FILE INPUT
with open("documents/document.txt",'r') as sample:
    print(sample.read())

# EXPECTED FILE OUTPUT
print(f"Total Lines: {len(load)} Total Words: {length}")
```

#### *Task 9: Empty Line Remover*

```python
# SAMPLE FILE
with open("documents/messy.txt",'r') as file_1:
    load = file_1.readlines()

with open("documents/clean.txt",'w') as file_2:
    for line in load:
        if not line.isspace():
            file_2.write(line)
# SAMPLE FILE
with open("documents/messy.txt",'r') as sample:
    print(sample.read())

print(f"{'-'*5} Output {'-'*5}")

# EXPECTED FILE OUTPUT
with open("documents/clean.txt",'r') as sample_1:
    print(sample_1.read())
```

#### *Task 10: Multi-Student Score Logger*

```python
with open("documents/scores.txt",'w') as file:
    file.write("Name, Score")

n = int(input("No.of students record : "))
name = None
score = None
def name():
    x = input("Enter Name : ").title()
    if x:
        return x
    else:
        print("Name can't be empty!")
        name()

def score():
    x = float(input("Enter Score : "))
    if x:
        return x
    else:
        print("Score can't be empty!")
        score()

with open("documents/scores.txt",'a') as w_file:
    for _ in range(n):
        w_file.write(f"\n{name()}, {score()}")

# EXPECTED FILE OUTPUT
with open("documents/scores.txt",'r') as sample:
    print(sample.read())
```
