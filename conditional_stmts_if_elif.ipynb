#### 1. Largest numbers among three integers

```python
a = int(input("Enter First value : "))
b = int(input("Enter Second value : "))
c = int(input("Enter Third value : "))
if a > b and a > c:
    print(f"largest is {a}")
elif b > a and b > c:
    print(f"largest is {b}")
else:
    print(f"largest is {c}")
```

#### 3. Check if a given number is greater than 0, if yes then print 'Positive'. If the given number is lesser than 0, then print 'Negative'. If the given number is exactly equal to 0, then print 'Zero'.

```python
value = int(input("Enter value : "))
if value > 0:
    print(f"{value} is positive")
elif value == 0:
    print(f"{value} is zero")
else:
    print(f"{value} is negative")
```

#### 4. A Library charges fine for books returned for late.

```python
days = int(input("Enter the number of books were returned late : "))
if (days >= 0) and (days <=5):
    fine = days * 40
    print(f"The Book were returned for {days} days late and fine amount charges 40 paisa per day")
    print(f"The total fine amount is {fine} paisa")
elif (days > 5) and (days <=10):
    fine = days * 65
    print(f"The Book were returned for {days} days late and fine amount charges 65 paisa per day")
    print(f"The total fine amount is {fine} paisa")
else:
    fine = days * 80
    print(f"The Book were returned for {days} days late and fine amount charges 80 paisa per day")
    print(f"The total fine amount is {fine} paisa")
```
