#### 3. Check if a given number is greater than 0, if yes then print 'Positive'. If the given number is lesser than 0, then print 'Negative'. If the given number is exactly equal to 0, then print 'Zero'.

```python
value = int(input("Enter value : "))
if value > 0:
    print(f"{value} is positive")
elif value == 0:
    print(f"{value} is zero")
else:
    print(f"{value} is negative")
```

#### 4. A Library charges fine for books returned for late.

```python
days = int(input("Enter the number of books were returned late : "))
if (days > 0) and (days <=5):
    fine = days * 40
    print(f"The Book were returned for {days} days late and fine amount charges 40 paisa per day")
    print(f"The total fine amount is {fine} paisa")
elif (days > 5) and (days <=10):
    fine = days * 65
    print(f"The Book were returned for {days} days late and fine amount charges 65 paisa per day")
    print(f"The total fine amount is {fine} paisa")
else:
    fine = days * 80
    print(f"The Book were returned for {days} days late and fine amount charges 80 paisa per day")
    print(f"The total fine amount is {fine} paisa")
```

#### 5. A basic calculator

```python
print("These are the Mathematical operations ( + , - , * , / )")
num_1 = int(input("Enter first value : "))
num_2 = int(input("Enter second value : "))
operator = input("Select the Operator : ")
if operator == '+':
    print(num_1 + num_2)
elif operator == '-':
    print(num_1 - num_2)
elif operator == '*':
    print(num_1 * num_2)
elif operator == '/':
    if num_1 != 0:
        print(num_1 / num_2)
    else:
        print("It can't be divisible by zero")
else:
    print(f"The given operator {operator} doesn't match with these ( + , - , * , / ) operators")

```

#### 8. The price of Laptop

```python
price = int(input("Price of Laptop : "))
if price <= 50000:
    print("You have no Discount")
    print(f"Total price is {price}Rs/-  for Laptop")
elif (price > 50000) and (price < 100000) :
    price -= price/100 * 10
    print("You have 10% Discount on the list Price")
    print(f"Total price is {price} Rs/-  for Laptop")
elif (price > 100000) and (price < 150000) :
    price -= price/100 * 15
    print("You have 15% Discount on the list Price")
    print(f"Total price is {price} Rs/-  for Laptop")
else:
    price -= price/100 * 20
    print("You have 20% Discount on the list Price")
    print(f"Total price is {price} Rs/-  for Laptop")
```
