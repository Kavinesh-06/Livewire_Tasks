## **NUMBER LOGIC & FUNCTIONS**
#### *1. Reverse a Number*


```python
def reverse(x):
    r = 0
    while x > 0:
        a = x%10
        r = r*10 + a
        x //= 10
    return r
value = int(input("Enter an value : "))
print(f"Input value : {value} \nOutput value : {reverse(value)}")
```

#### *2. Sum of Digits*

```python
def sum_of(x):
    s = 0
    while x>0:
        r = x%10
        s += r
        x //= 10
    return s



Ids = int(input("Enter an value : "))
print(f"The sum of its digits : {sum_of(Ids)}")
```

#### *3. Prime Number Checker*

```python
def is_prime(value):
    flag = True
    for i in range(2,value//2 +1):
        if value % i == 0:
            flag = False
            break
    if flag:
        return "Prime"
    else:
        return "Not Prime"
print(is_prime(29))
```

#### *4. Prime Numbers in a Range*


```python
def is_prime(start,end):
    for j in range(start,end+1):
        flag = True
        for i in range(2,j//2 +1):
            if j % i == 0:
                flag = False
                break
        if flag:
            print(j,end=' ')
is_prime(10,30)
```

#### *5. Palindrome Number*

```python
def is_palindrome(x):
    r = 0
    y = x
    while x > 0:
        a = x%10
        r = r*10 + a
        x //= 10
    if r == y:
        print("Palindrome")
    else:
        print("Not Palindrome")
is_palindrome(1221)
```

## **STRING-BASED FUNCTIONS** 
#### *6. Reverse a String*


```python
def reverse_string(x):
    y = len(x)-1
    rev = ""
    for i in range(y,-1,-1):
        rev += x[i]
    return(rev)
print(reverse_string("programming"))
```

#### *7. Count Vowels*

```python
def count_vowels(x):
    vowel = "aeiou"
    count = 0
    for i in x:
        if i in vowel:
            count += 1
    print(count)
a = input("Enter sentence : ").lower()
count_vowels(a)
```

#### *8. Character Frequency*

```python
def freq(x):
    y = ""
    for i in x:
        if i not in y:
            print(f"{i} : {x.count(i)}")
        y += i

freq("banana")
```

#### *9. First Non-Repeating Character*


```python
def non_repeat(x):
    for i in x:
        if x.count(i) == 1:
            print(i)
            break
a = input("Enter some words : ")
non_repeat(a)
```

#### *10. Remove Duplicate Characters*


```python
def remove_duplicates(x):
    a = ""
    for i in x:
        if i not in a:
            a += i
    print(a)
remove_duplicates("programming")
```

## **LIST + FUNCTION**
#### *11. Find the Second Largest Number*

```python
def sec_large(x):
    high = x[0]
    second = x[0]
    for i in x:
        if i > high:
            second = high
            high = i
        if second < i < high:
            second = i

    print(second)
a = [10, 45, 23, 67, 34]   
sec_large(a)   
```
