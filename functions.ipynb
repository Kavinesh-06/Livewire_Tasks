## **NUMBER LOGIC & FUNCTIONS**
#### *1. Reverse a Number*


```python
def reverse(x):
    r = 0
    while x > 0:
        a = x%10
        r = r*10 + a
        x //= 10
    return r
value = int(input("Enter an value : "))
print(f"Input value : {value} \nOutput value : {reverse(value)}")
```

#### *2. Sum of Digits*

```python
def sum_of(x):
    s = 0
    while x>0:
        r = x%10
        s += r
        x //= 10
    return s



Ids = int(input("Enter an value : "))
print(f"The sum of its digits : {sum_of(Ids)}")
```

#### *3. Prime Number Checker*

```python
def is_prime(value):
    flag = True
    for i in range(2,value//2 +1):
        if value % i == 0:
            flag = False
            break
    if flag:
        return "Prime"
    else:
        return "Not Prime"
print(is_prime(29))
```

#### *4. Prime Numbers in a Range*


```python
def is_prime(start,end):
    for j in range(start,end+1):
        flag = True
        for i in range(2,j//2 +1):
            if j % i == 0:
                flag = False
                break
        if flag:
            print(j,end=' ')
is_prime(10,30)
```

#### *5. Palindrome Number*

```python
def is_palindrome(x):
    r = 0
    y = x
    while x > 0:
        a = x%10
        r = r*10 + a
        x //= 10
    if r == y:
        print("Palindrome")
    else:
        print("Not Palindrome")
is_palindrome(1221)
```

## **STRING-BASED FUNCTIONS** 
#### *6. Reverse a String*


```python
def reverse_string(x):
    y = len(x)-1
    rev = ""
    for i in range(y,-1,-1):
        rev += x[i]
    return(rev)
print(reverse_string("programming"))
```

#### *7. Count Vowels*

```python
def count_vowels(x):
    vowel = "aeiou"
    count = 0
    for i in x:
        if i in vowel:
            count += 1
    print(count)
a = input("Enter sentence : ").lower()
count_vowels(a)
```

#### *8. Character Frequency*

```python
def freq(x):
    y = ""
    for i in x:
        if i not in y:
            print(f"{i} : {x.count(i)}")
        y += i

freq("banana")
```

#### *9. First Non-Repeating Character*


```python
def non_repeat(x):
    for i in x:
        if x.count(i) == 1:
            print(i)
            break
a = input("Enter some words : ")
non_repeat(a)
```

#### *10. Remove Duplicate Characters*


```python
def remove_duplicates(x):
    a = ""
    for i in x:
        if i not in a:
            a += i
    print(a)
remove_duplicates("programming")
```

## **LIST + FUNCTION**
#### *11. Find the Second Largest Number*

```python
def sec_large(x):
    high = x[0]
    second = x[0]
    for i in x:
        if i > high:
            second = high
            high = i
        if second < i < high:
            second = i

    print(second)
a = [10, 45, 23, 67, 34]   
sec_large(a)   
```

#### *12. Remove Duplicate Elements from a List*

```python
def remove_duplicate(x):
    y = []
    for i in x:
        if i not in y:
            y.append(i)
    return y


a = [10, 20, 10, 30, 20, 40]
print(remove_duplicate(a))
```

#### *13. Move All Zeros to the End*

```python
def move_0(x):
    for i in range(len(x)):
        for j in range(i+1,len(x)):
            if x[i] == 0:
                x[i],x[j] = x[j],x[i]
    return x
a = [0, 5, 0, 3, 8, 0, 2]
print(move_0(a))
```

#### *14. Find the Missing Number*

```python
def missing_no(x):
    for i in range(x[0],x[-1]):
        if i not in x:
            return i
a = [11, 12, 13, 15, 16]
print(f"""Natural numbers(N) : {a[-1]}
Data-entry : {a}
Missing value : {missing_no(a)}""")
```

#### *15. Find Pairs With Given Sum*

```python
def pairs(x,key):
    y = []
    for i in x:
        m = key - i
        if m in x and m > 0 and i not in y:
            y.append(m)
            print(i,m)

value = [2, 7, 11, 15, 3, 6]
target = 9
pairs(value,target)
```

## **ADVANCED ZOHO-STYLE LOGIC**
#### *16. Anagram Without Sorting*


```python
def is_anagram(x,y):
    check = ""
    if len(x) == len(y):
        for i in x:
            if i in y:
                check += i
            else:
                return (f"{x} and {y} is not an Anagram")
        if check == x:
            return (f"{x} and {y} is Anagram")
    else:
        return (f"{x} and {y} is not an Anagram")

a = input("Enter a text : ")
b = input("Enter a text : ")
print(is_anagram(a,b))
```

#### *17. String Compression*


```python
def string_compression(x):
    count = 1
    comp = ""
    for i in range(len(x)):
        if i < len(x)-1 and x[i] == x[i+1]:
            count += 1
        else:
            comp += x[i] + str(count)
            count = 1
    return comp
a = "aaabbcccc"
print(string_compression(a))
```

#### *18. Find Leaders in an Array*

```python
def find_leader(a):
    value = [a[-1]]
    for i in range(len(a)-1,-1,-1):
        if a[i] > value[-1]:
            value.append(a[i])
    return value[::-1]

a = [16, 17, 4, 3, 5, 2]
print(find_leader(a))
```

#### *19. Employee Salary Ranking*

```python
def salary_ranking(salary):
    high = salary[0]
    second = salary[0]
    for i in salary:
        if i[1] > high[1]:
            second = high
            high = i
        if second[1]<i[1]<high[1]:
            second = i
    return (f"Highest : {high}\nSecond : {second}")

salary = (('Arun', 45000),('Priya', 52000),('Kumar', 48000),('Divya', 60000))
print(f"Highest : {salary_ranking(salary)}")
```

#### *20. Student Result Processing*

```python
def calculate_result(x):
    total = sum(x)
    average = total/5
    Grade = None
    if average >= 40:
        result = 'Pass'
        if 90 <= average <= 100:
            Grade = 'A'
        elif 80 <= average <= 89:
            Grade = 'B'
        elif 70 <= average <= 79:
            Grade = 'C'
        elif 60 <= average <= 69:
            Grade = 'D'
        else:
            Grade = 'F' 
    else:
        result = 'Fail'
    return(f"Total : {total} \nAverage : {average} \nGrade : {Grade} \nResult : {result}")
a = []
for _ in range(5):
    b = int(input("Enter Mark : "))
    a.append(b)
print(calculate_result(a))
```
