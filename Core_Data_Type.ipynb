## **A . List**

#### 1. Linear Search

```python
List = [101, 205, 304, 405, 502]
Target = 405
a = False
for x in List:
    if x == Target:
        print(f"Element found at index of {List.index(x)}")
        a = False
        break
    else:
        a = True
if a:
    print("Element not founded")
```

#### 2. Find Second Largest

```python
list_1 = [45, 12, 89, 95, 67, 34, 90]
high = list_1[0]
second = list_1[0]
for x in list_1:
    if x > high:
        high = x
    if second < x < high:
        second = x
print(f"First Highest value in the List is {high}")
print(f"Second Highest value in the List is {second}")
```

#### 3. Find Second Smallest

```python
list_1 = [45, 12, 89, 95, 67, 34, 90]
high = list_1[0]
second = list_1[0]
for x in list_1:
    if x < high:
        high = x
    if second > x > high:
        second = x
print(f"First Lowest value in the List is {high}")
print(f"Second Lowest value in the List is {second}")
```

#### 4. Sort Without Built-in Function

```python
list_1 = [64, 34, 25, 12, 22, 11, 90]
for i in range(len(list_1)):
    for j in range(i+1,len(list_1)):
        if list_1[i] > list_1[j] :
            list_1[i],list_1[j] = list_1[j],list_1[i]

print(f"Sorting a list of numbers in ascending order : {list_1}")
```

#### 5. Remove Duplicates

#### 8. Move Zeros

```python
a = [0, 5, 0, 3, 8, 0, 2]
for x in range(len(a)):
    for y in range(x+1,len(a)):
        if a[x] == 0:
            a[x],a[y] = a[y],a[x]
print(f"Move all zeros to the end : {a}")
```

#### 12. Tuple Sorting

```python
x = (50, 10, 40, 20, 30)
x1 = ()
y = ()
for n in range(len(x)):
    x1 += (x[n],)
for i in range(len(x)):
    small = x[0]
    index = 0
    for j in range(len(x)):
        if small > x[j]:
            small = x[j]
            index = j
    y += (small,)
    x = x[:index] + x[index+1:]  
print(f"The original tuple : {x1}")    
print(f"Tuple sorting without changing tuple data types and built-in function : {y}")
```

#### 33. Sort Dictionary by Values

```python
employee = {"Arun": 45000, "Priya": 52000, "Kumar": 48000, "Divya": 60000}
x = {}
keys = list(employee.keys())
for i in range(len(keys)):
    low = None
    for j in range(len(keys)):
        if keys[j] in employee:
            if low is None or employee[keys[j]] < employee[low]:
                low = keys[j]
    x[low] = employee[low]
    del employee[low]
print(x)
```

#### 40. Student Ranking 

```python
student = {"Arun": 85, "Priya": 92, "Kumar": 78, "Divya": 95, "Rahul": 88}
sort_1 = sorted(student,key=student.get,reverse="True")
x = {}
top_3 = []
count = 0
for k in sort_1:
    x[k] = student[k]
for i in x.items():
    if count == 3:
        break
    top_3.append(i)
    count += 1
y = list(x.items())
print(f"""
Highest : {y[0]},
Lowest : {y[-1]},
sorted : {x},
Top 3 : {top_3}
""")
```
